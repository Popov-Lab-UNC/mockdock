```python
jan14_run = pd.read_csv('benchmark_runs/run_20260114_082857/run_summary.csv')
jan14_run[(jan14_run['success_rate_rmsd'] > 80.0) & (jan14_run['n_compounds_matched_2d'] > 20.0)].sort_values(by='rmsd_pearson', ascending=True).head(50)
```

```python
import matplotlib.pyplot as plt

# Filter and sort the dataframe as described
sorted_df = jan14_run[(jan14_run['success_rate_rmsd'] > 80.0) & (jan14_run['n_compounds_matched_2d'] > 20.0)].sort_values(by='rmsd_pearson', ascending=True).head(50)

plt.plot(sorted_df.reset_index(drop=True).index, sorted_df['rmsd_pearson'], label='rmsd_pearson')
plt.plot(sorted_df.reset_index(drop=True).index, sorted_df['rmsd_spearman'], label='rmsd_spearman')
plt.plot(sorted_df.reset_index(drop=True).index, sorted_df['rmsd_r2'], label='rmsd_r2')
plt.legend()
plt.show()
```

```python
sorted_df[sorted_df['rmsd_spearman'] < -0.6].to_csv('best_pl_systems.csv', index=False)
```
